```python
import pandas as pd
```

```python
def split_names_tan(df):
    df.loc[:, "first_name"] = df.Name.str.replace(r"(\w+) (\w+)", r"\1", regex=True)
    df.loc[:, "last_name"] = df.Name.str.replace(r"(\w+) (\w+)", r"\2", regex=True)
    return df 

def split_names_iberia(df):
    df.loc[:, "first_name"] = df.Name.str.replace(r"(\w+)\, (\w+)", r"\2", regex=True)
    df.loc[:, "last_name"] = df.Name.str.replace(r"(\w+)\, (\w+)", r"\1", regex=True)
    return df 

def read_tbls():
    dfa = pd.read_csv("../data/input/brady_baton_rouge_da_2018.csv")
    dfb = pd.read_csv("../data/input/brady_baton_rouge_da_2021.csv")
    dfc = pd.read_csv("../data/input/brady_morehouse_da_2022.csv")
    dfd = pd.read_csv("../data/input/brady_ouachita_da_2021.csv")

    dfe = pd.read_csv("../data/input/tangipahoa_da_brady_2021.csv")
    dfe["source_agency"] = "tangipahoa da"
    dfe = dfe.pipe(split_names_tan)


    dff = pd.read_csv("../data/input/brady_iberia_da_2022.csv")
    dff = dff.pipe(split_names_iberia)
    dff["source_agency"] = "iberia da"

    dfg = pd.read_csv("../data/input/brady_caddo.csv")
    
    df = pd.concat([dfa, dfb, dfc, dfd, dfe, dff, dfg])

    df = df[["last_name", "middle_name", "first_name", "agency", "source_agency"]]

    df.loc[:, "last_name"] = df.last_name.str.capitalize()

    df.loc[:, "first_name"] = df.first_name.str.capitalize()

    df.loc[:, "middle_name"] = df.middle_name.str.capitalize()
    return df 


df = read_tbls()

df
df.to_csv("../data/output/brady.csv", index=False)
```
