```python
import pandas as pd
```

```python
def read_tbl():
    df = pd.read_excel("../data/input/INC0261359.xlsx", sheet_name="Employment")
    df = df.rename(columns={"Fullname": "name",
                            "Employeeid": "person_nbr",
                            "BirthYear": "year_of_birth",
                            "AgencyName": "agency_name",
                            "AgencyORINumber": "ori_number",
                            "StartDate": "hire_date",
                            "EndDate": "end_date", 
                            "Status": "employment_status",
                            "SeparationReason": "separation_reason"})
    return df 

def normalize_birth_year(df):
    df.loc[:, "year_of_birth"] = df.year_of_birth.fillna("").astype(str).str.replace(r"\.(.+)", "", regex=True)
    return df 

def clean_dates(df):
    df.loc[:, "hire_date"] = df.hire_date.astype(str).str.replace(r"(\w{2}):(\w{2}):(\w{2})", "", regex=True)
    
    df.loc[:, "end_date"] = df.end_date.astype(str).str.replace(r"(\w{2}):(\w{2}):(\w{2})", "", regex=True)
    return df


def parse_names(df):
    names = df.name.fillna("").astype(str).str.lower().str.strip().str.replace(r"\n", " ").str.extract(r"^(\w+)\, (\w{2,3}\.)? ?(\w+) ?(\w+)?\.?$")

    df.loc[:, "last_name"] = names[0]
    df.loc[:, "suffix"] = names[1]
    df.loc[:, "first_name"] = names[2]
    df.loc[:, "middle_name"]= names[3]
    return df


def clean_sep_reason(df):
    df.loc[:, "separation_reason"] = (df.separation_reason
                                        .fillna("")
                                        .str.lower()
                                        .str.strip()
                                        .str.replace(r" - ", " ", regex=True)
                                        .str.replace(r"retirement retirement", "retirement", regex=False)
                                        .str.replace(r"retirement service retirement", "retirement", regex=False)
    )
    return df

df = read_tbl()

df = df.pipe(normalize_birth_year).pipe(clean_dates).pipe(parse_names).pipe(clean_sep_reason)

df.separation_reason.value_counts()
```
