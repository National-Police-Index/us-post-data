```python
def clean_person_nbrs(df):
    df.loc[:, "person_nbr"] = df.person_nbr.astype(str).str.replace(
        r"\s+", "", regex=True
    )
    return df
```

```python
def filter_agency_data(df):
    df = df.rename(columns={"agcy_name": "agency_name"})

    df = df[["agency_name", "agcy_nbr"]]
    return df


df_agencies = df_agencies.pipe(filter_agency_data)
```

```python
df_employment_history = pd.merge(df_employment, df_person_data, on="person_nbr")

df_employment_history = df_employment_history.sort_values("person_nbr")

df_employment_history.head(30)

df_employment_history = df_employment_history[
    [
        "first_name",
        "middle_name",
        "last_name",
        "sex_code",
        "race_code",
        "birth_year",
        "employ_start_date",
        "separation_date",
        "separation_code",
        "person_nbr",
        "employment_nbr",
        "agcy_nbr",
    ]
]


def rename_cols(df):
    df = df.rename(
        columns={
            "sex_code": "sex",
            "race_code": "race",
            "birth_year": "year_of_birth",
            "employ_start_date": "start_date",
            "separation_date": "end_date",
            "separation_code": "separation_reason",
        }
    )
    return df


def clean_demo_data(df):

    df.loc[:, "sex"] = (
        df.sex.str.lower()
        .str.strip()
        .fillna("")
        .str.replace(r"^f$", "Female", regex=True)
        .str.replace(r"^m$", "Male", regex=True)
        .str.replace(r"(u|o)", "", regex=True)
    )

    df.loc[:, "race"] = (
        df.race.str.lower()
        .str.strip()
        .fillna("")
        .str.replace(r"^(his|h)$", "Hispanic", regex=True)
        .str.replace(r"^(wh|whi)$", "White", regex=True)
        .str.replace(r"^blk$", "Black", regex=True)
        .str.replace(r"^as$", "Asian", regex=True)
        .str.replace(r"(oth)", "Other", regex=True)
        .str.replace(r"na", "", regex=True)
    )

    return df


def clean_birth_year(df):
    df.loc[:, "year_of_birth"] = (
        df.year_of_birth.astype(str)
        .str.replace(r"\.(.+)", "", regex=True)
        .str.replace(r"\s+", "", regex=True)
    )
    return df


def clean_dates(df):
    df.loc[:, "start_date"] = (
        df.start_date.astype(str)
        .str.replace(r"(\w{2}):(\w{2}):(\w{2})", "", regex=True)
        .str.replace(r"\s+", "", regex=True)
    )
    df.loc[:, "end_date"] = (
        df.end_date.astype(str)
        .str.replace(r"(\w{2}):(\w{2}):(\w{2})", "", regex=True)
        .str.replace(r"\s+", "", regex=True)
        .str.replace(r"nan", "", regex=False)
    )
    return df


def merge_employment_history_w_agencies(dfa, dfb):
    df = pd.merge(dfa, dfb, on="agcy_nbr")
    return df


df_employment_history = (
    df_employment_history.pipe(rename_cols)
    .pipe(clean_demo_data)
    .pipe(clean_birth_year)
    .pipe(clean_dates)
)

# agency names aren't clean. plz clean @claudina
df_employment_history = merge_employment_history_w_agencies(
    df_employment_history, df_agencies
)
```
